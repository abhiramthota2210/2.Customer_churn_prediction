# Logistic Regression - Mathematical Implementation

Suppose the training dataset contains:

- **m** training examples
- **n** input features

## Step 1: Training Data

Feature matrix:

\[
X=
\begin{bmatrix}
x_1^{(1)} & x_2^{(1)} & \cdots & x_n^{(1)}\\
x_1^{(2)} & x_2^{(2)} & \cdots & x_n^{(2)}\\
\vdots & \vdots & \ddots & \vdots\\
x_1^{(m)} & x_2^{(m)} & \cdots & x_n^{(m)}
\end{bmatrix}
\]

Target vector:

\[
y=
\begin{bmatrix}
0\\
1\\
0\\
\vdots
\end{bmatrix}
\]

---

## Step 2: Initialize Parameters

Weight vector:

\[
\theta=
\begin{bmatrix}
\theta_1\\
\theta_2\\
\vdots\\
\theta_n
\end{bmatrix}
\]

Bias:

\[
b
\]

Initially,

\[
\theta = 0,\qquad b = 0
\]

---

## Step 3: Linear Combination

For every training example,

\[
z^{(i)}
=
\theta_1x_1^{(i)}
+\theta_2x_2^{(i)}
+\cdots
+\theta_nx_n^{(i)}
+b
\]

Vectorized form:

\[
z = X\theta + b
\]

---

## Step 4: Sigmoid Function

The sigmoid converts the linear output into a probability.

\[
\hat y
=
\sigma(z)
=
\frac{1}{1+e^{-z}}
\]

Example:

If

\[
z=2
\]

then

\[
\hat y
=
\frac1{1+e^{-2}}
=
0.88
\]

Interpretation:

\[
P(y=1)=0.88
\]

---

## Step 5: Binary Cross Entropy Loss

Loss for one training example:

\[
L
=
-y\log(\hat y)
-
(1-y)\log(1-\hat y)
\]

Examples:

If

\[
y=1,\quad \hat y=0.95
\]

Loss is very small.

If

\[
y=1,\quad \hat y=0.05
\]

Loss becomes very large.

---

## Step 6: Cost Function

Average loss over all training examples:

\[
J(\theta,b)
=
-\frac1m
\sum_{i=1}^{m}
\left[
y^{(i)}
\log(\hat y^{(i)})
+
(1-y^{(i)})
\log(1-\hat y^{(i)})
\right]
\]

Goal:

\[
\boxed{\text{Minimize } J(\theta,b)}
\]

---

## Step 7: Compute Gradients

Gradient with respect to weights:

\[
\frac{\partial J}{\partial\theta}
=
\frac1m
X^T(\hat y-y)
\]

Gradient with respect to bias:

\[
\frac{\partial J}{\partial b}
=
\frac1m
\sum(\hat y-y)
\]

---

## Step 8: Gradient Descent

Update the parameters.

Weights:

\[
\theta
=
\theta
-
\alpha
\frac{\partial J}{\partial\theta}
\]

Bias:

\[
b
=
b
-
\alpha
\frac{\partial J}{\partial b}
\]

where

- \(\alpha\) = Learning Rate

---

## Step 9: Repeat

Repeat until convergence:

1. Compute \(z = X\theta + b\)
2. Compute \(\hat y = \sigma(z)\)
3. Compute the cost
4. Compute the gradients
5. Update \(\theta\) and \(b\)

---

## Step 10: Prediction

For a new sample,

\[
z=x\theta+b
\]

Compute probability:

\[
\hat y
=
\frac1{1+e^{-z}}
\]

Classification rule:

\[
\hat y \ge 0.5
\Rightarrow
1
\]

\[
\hat y < 0.5
\Rightarrow
0
\]

---

# Complete Pipeline

```
Input Features (X)
        │
        ▼
Linear Model
z = Xθ + b
        │
        ▼
Sigmoid Function
ŷ = 1 / (1 + e^(-z))
        │
        ▼
Binary Cross Entropy Loss
        │
        ▼
Compute Gradients
        │
        ▼
Gradient Descent Update
θ = θ − α dθ
b = b − α db
        │
        ▼
Repeat Until Convergence
        │
        ▼
Predict on Test Data
```

## Difference from Linear Regression

| Linear Regression | Logistic Regression |
|-------------------|---------------------|
| Continuous Output | Binary Output |
| Identity Function | Sigmoid Function |
| Mean Squared Error (MSE) | Binary Cross Entropy (Log Loss) |
| Linear Prediction | Probability Prediction |
